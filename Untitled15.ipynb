## Planteamiento del Problema para los Estudiantes: Un hospital quiere implementar un modelo de Inteligencia Artificial para detectar si un paciente tiene riesgo de diabetes antes de realizar análisis complejos. Les entregan un conjunto de datos con las siguientes columnas: [id_paciente, edad, nivel_glucosa, presion_arterial, indice_masa_corporal, diagnostico_diabetes] Instrucciones para la actividad:

### Identifiquen cuál es el objetivo ( y ).
### Identifiquen cuáles son las características ( incógnita ).
### ¿Existe alguna columna que debamos eliminar por no aportar información útil al aprendizaje? Justifiquen su respuesta.

```python
## Ejepmlo a dar

import pandas as pd

# 1. Crear un dataset de ejemplo simular a un CSV
datos = {
    'distancia_km': [2.5, 4.0, 1.2, 5.8, 3.1],
    'trafico_nivel': [1, 3, 1, 3, 2],        # 1: Bajo, 2: Medio, 3: Alto
    'edad_repartidor': [22, 35, 19, 28, 40],
    'tiempo_entrega_min': [15, 32, 10, 42, 25] # Lo que queremos predecir
}

df = pd.DataFrame(datos)

# 2. Separar Variables de Entrada (X) y Variable Objetivo (y)
X = df[['distancia_km', 'trafico_nivel', 'edad_repartidor']] # Features / Entradas
y = df['tiempo_entrega_min']                                  # Target / Salida

# 3. Mostrar estructura
print("--- DATOS DE ENTRADA (FEATURES - X) ---")
print(X.head(2))

print("\n--- VARIABLE OBJETIVO (TARGET - y) ---")
print(y.head(2))
```

```python
## Trabajo de Hospital

import pandas as pd

# 1. Crear un dataset de ejemplo similar a un CSV del hospital
datos = {
    'id_paciente': [101, 102, 103, 104, 105],
    'edad': [45, 23, 60, 35, 52],
    'nivel_glucosa': [120, 85, 200, 90, 150],
    'presion_arterial': [80, 70, 95, 75, 85],
    'indice_masa_corporal': [28.4, 21.2, 33.1, 24.5, 30.0],
    'diagnostico_diabetes': [1, 0, 1, 0, 1]  # 1: Sí, 0: No (Lo que queremos predecir)
}

df = pd.DataFrame(datos)

# 2. Separar Variables de Entrada (X) y Variable Objetivo (y)
# Nota: Excluimos 'id_paciente' porque no aporta información útil para el aprendizaje.
X = df[['edad', 'nivel_glucosa', 'presion_arterial', 'indice_masa_corporal']] # Features / Entradas
y = df['diagnostico_diabetes']                                            # Target / Salida

# 3. Mostrar estructura
print("--- DATOS DE ENTRADA (FEATURES - X) ---")
print(X.head(2))

print("\n--- VARIABLE OBJETIVO (TARGET - y) ---")
print(y.head(2))
```
